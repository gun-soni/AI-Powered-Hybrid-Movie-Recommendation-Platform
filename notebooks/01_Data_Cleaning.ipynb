```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
```

```python
movies = pd.read_csv("sample_data/movies_metadata.csv", low_memory=False)

credits = pd.read_csv("sample_data/credits.csv")

keywords = pd.read_csv("sample_data/keywords.csv")

ratings = pd.read_csv("sample_data/ratings.csv")

links = pd.read_csv("sample_data/links.csv")
```

```python
print("Movies:", movies.shape)
print("Credits:", credits.shape)
print("Keywords:", keywords.shape)
print("Ratings:", ratings.shape)
print("Links:", links.shape)
```

### Step 11 & 12 — Remove Invalid IDs and Convert Data Types

```python
movies = movies[movies["id"].astype(str).str.isnumeric()].copy()
movies["id"] = movies["id"].astype(int)
credits["id"] = credits["id"].astype(int)
keywords["id"] = keywords["id"].astype(int)
```

### Step 13 — Merge Dataset

```python
movies = movies.merge(
    credits,
    on="id"
)

movies = movies.merge(
    keywords,
    on="id"
)
```

### Step 14 — Verify Merge

```python
movies.shape
```

### Step 15 — Keep Only Useful Columns

```python
movies = movies[
    [
        "id",
        "title",
        "overview",
        "genres",
        "cast",
        "crew",
        "keywords",
        "vote_average",
        "vote_count",
        "popularity",
        "release_date",
        "original_language"
    ]
]
```

### Step 16 — Remove Missing Values

```python
movies.dropna(inplace=True)
movies.reset_index(drop=True, inplace=True)
```

### Step 17 — Verify Final Dataset

```python
movies.info()
```

### Step 18 — Save Dataset

```python
movies.to_csv(
    "clean_movies.csv",
    index=False
)
```
