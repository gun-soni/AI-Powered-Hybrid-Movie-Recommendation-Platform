```python
import pandas as pd

from surprise import Dataset

from surprise import Reader

from surprise import SVD

from surprise import accuracy

from surprise.model_selection import train_test_split

import pickle
```

```python
ratings = pd.read_csv(
    "ratings.csv"
)

ratings.head()
```

```python
ratings = ratings[
    [
        "userId",
        "movieId",
        "rating"
    ]
]
```

```python
ratings.shape
```

```python
ratings = ratings.sample(
    n=1000000,
    random_state=42
)
```

```python
reader = Reader(
    rating_scale=(0.5,5)
)
```

```python
data = Dataset.load_from_df(

    ratings,

    reader

)
```

```python
trainset,testset = train_test_split(

    data,

    test_size=0.2,

    random_state=42

)
```

```python
model = SVD(

    n_factors=100,

    lr_all=0.005,

    reg_all=0.02,

    random_state=42

)
```

```python
model.fit(

    trainset

)
```

```python
predictions = model.test(

    testset

)
```

```python
accuracy.rmse(

    predictions
)
```

```python
pickle.dump(

    model,

    open(

        "svd_model.pkl",

        "wb"

    )

)
```
